# SpaceX Falcon 9 First Stage Landing Prediction

## Lab 3: Data Wrangling

In this lab, we will perform some Exploratory Data Analysis (EDA) to find patterns in the data and determine the label for training supervised models. The dataset contains several cases where the booster did not land successfully — sometimes a landing was attempted but failed due to an accident. We will mainly convert the landing `Outcome` strings into training labels, where `1` means the booster successfully landed and `0` means it was unsuccessful.

## Objectives

- Perform exploratory Data Analysis and determine training labels
- Calculate the number of launches on each site, and the number and occurrence of each orbit and mission outcome per orbit type
- Create a landing outcome label from the `Outcome` column
- Export the labeled dataset to `dataset_part_2.csv`

```python
import pandas as pd
import numpy as np
```

### Load the dataset

We load the SpaceX dataset collected in Lab 1 (also hosted on the course object storage) into a pandas DataFrame.

```python
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"
df = pd.read_csv(URL)
df.head(10)
```

### Identify and calculate the percentage of missing values

First, we check which attributes contain missing values. The `LandingPad` column has many NaN values — these correspond to launches where no landing pad was used (the booster was expended or landed in the ocean). We also verify the data types of every column.

```python
df.isnull().sum() / df.shape[0] * 100
```

```python
df.dtypes
```

### Calculate the number of launches on each site

Using `value_counts` on the `LaunchSite` column, we can see that the Cape Canaveral Space Launch Complex 40 (CCAFS SLC-40) hosted the most launches, followed by Kennedy Space Center LC-39A and Vandenberg SLC-4E.

```python
df['LaunchSite'].value_counts()
```

### Calculate the number and occurrence of each orbit

Each launch targets a specific orbit. The most common orbits are GTO (geostationary transfer orbit), ISS (low Earth orbit used to reach the International Space Station), VLEO (very low Earth orbit) and PO (polar orbit).

```python
df['Orbit'].value_counts()
```

### Calculate the number and occurrence of mission outcome per orbit type

The `Outcome` column combines the landing success boolean with the landing type, e.g. `True ASDS` means the booster successfully landed on a drone ship (Autonomous Spaceport Drone Ship), `False RTLS` means a return-to-launch-site landing failed, and `None None` means no landing was attempted.

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

### Create a landing outcome label from the `Outcome` column

We define the set of **bad outcomes** — `False ASDS`, `False Ocean`, `False RTLS`, `None ASDS` and `None None` — and map every outcome to a class label: `0` for a bad outcome and `1` for a successful landing. This `Class` column is the label we will train our supervised models to predict.

```python
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
bad_outcomes
```

```python
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
df[['Class']].head(8)
```

### Success rate

Since every row is either a success (1) or a failure (0), the mean of the `Class` column gives the overall landing success rate of the dataset.

```python
df["Class"].mean()
```

```python
df.to_csv("dataset_part_2.csv", index=False)
```
