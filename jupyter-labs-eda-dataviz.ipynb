# SpaceX Falcon 9 First Stage Landing Prediction

## Lab 4: Exploratory Data Analysis with Visualization

In this lab, we perform exploratory data analysis on the labeled SpaceX dataset using seaborn and matplotlib. We visualize the relationships between flight number, payload mass, launch site, orbit and the landing outcome (`Class`), and observe the trend of the launch success rate over the years.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(URL)
df.head(5)
```

### FlightNumber vs. LaunchSite

We plot a scatter plot of `FlightNumber` vs. `LaunchSite` with the landing outcome (`Class`) as the hue. This lets us see how the success rate at each site evolved as SpaceX accumulated more flights.

```python
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()
```

**Observation:** Earlier flights (low flight numbers) failed more often, especially at CCAFS SLC-40, while recent flights are mostly successful. CCAFS SLC-40 has hosted the most launches, and KSC LC-39A shows a very high success rate on its flights.

### Payload vs. Launch Site

Next, we examine how the payload mass relates to the launch site and the landing outcome. If a payload is very heavy, the booster has less margin to return, so we expect to see patterns here.

```python
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()
```

**Observation:** VAFB SLC-4E has not launched payloads heavier than about 10,000 kg. Very heavy payloads (> 9,000 kg) landed successfully, while failures appear more frequently in the low-to-mid payload range at CCAFS SLC-40.

### Success rate of each orbit type

We now visualize the relationship between the landing success rate and the orbit type using a bar chart of the mean `Class` value per orbit.

```python
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
sns.barplot(x='Orbit', y='Class', data=orbit_success)
plt.xlabel("Orbit", fontsize=15)
plt.ylabel("Mean Success Rate", fontsize=15)
plt.show()
```

**Observation:** The orbits ES-L1, GEO, HEO and SSO have a 100% success rate, while SO has never landed successfully. VLEO has a high success rate with a decent number of flights, whereas GTO has the lowest success rate among the frequently used orbits (around 50%).

### FlightNumber vs. Orbit type

For each orbit, we plot `FlightNumber` vs. `Orbit` with the outcome hue. This shows whether the success rate for an orbit is related to the number of flights (experience).

```python
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()
```

**Observation:** In the LEO orbit, success appears related to the number of flights — later flights succeed more often. In the GTO orbit, on the other hand, there seems to be no relationship between flight number and success.

### Payload vs. Orbit type

We plot payload mass against orbit type with the outcome hue, to reveal how heavy payloads influence landing success on different orbits.

```python
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()
```

**Observation:** With heavy payloads, the successful landing rate is higher for Polar, LEO and ISS orbits. However, for GTO we cannot distinguish this well, as both successful and failed landings occur across the whole payload range.

### Launch success yearly trend

Finally, we extract the year from the `Date` column and plot a line chart of the average success rate per year to see the overall trend.

```python
year = []
def Extract_year(date):
    for i in df["Date"]:
        year.append(i.split("-")[0])
    return year

Extract_year(df["Date"])
df['Year'] = year
yearly_success = df.groupby('Year')['Class'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Class', data=yearly_success, marker='o')
plt.xlabel("Year", fontsize=15)
plt.ylabel("Success Rate", fontsize=15)
plt.title("Launch Success Yearly Trend", fontsize=18)
plt.show()
```

**Observation:** The success rate has been increasing since 2013, with a notable dip around 2018, and reached above 80% by 2019–2020. SpaceX's landing technology has clearly matured over time.

## Summary

The EDA shows that launch site, orbit type, payload mass and accumulated flight experience all relate to the landing outcome, and the success rate improved steadily year over year. These insights motivate the features used in the predictive modeling lab.
